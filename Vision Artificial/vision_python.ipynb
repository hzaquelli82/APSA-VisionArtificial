```python
import cv2
from ultralytics import YOLO

# Cargar el modelo preentrenado YOLO
model = YOLO("yolov8n.pt")

# URL del flujo de video RTSP
url_rtsp = "rtsp://admin:Apsa2812@192.168.0.209:554/(h264/MPEG-4)/ch(CANAL)/(main/sub)/av_stream"

# Crear un objeto de captura de video
cap = cv2.VideoCapture(url_rtsp)

# Verificar si la captura de video se ha iniciado correctamente
if not cap.isOpened():
    print("Error al iniciar la captura de video")
    exit()

while True:
    ret, frame = cap.read()
    if not ret:
        print("Error al leer un frame")
        break

    # Ejecutar el modelo YOLO en el frame
    detections = model(frame)[0]

    # Dibujar las detecciones en el frame
    for box in detections.boxes:
        # Extraer la etiqueta y la confianza
        label = model.names[int(box.cls)]
        confidence = float(box.conf.item())  # Convertir tensor a flotante

        # Filtrar detecciones débiles
        if confidence < 0.5:  # Ajusta el umbral según sea necesario
            continue

        # Obtener coordenadas del cuadro delimitador
        xmin, ymin, xmax, ymax = box.xyxy[0].tolist()  # Convertir tensor a lista

        # Dibujar el cuadro delimitador y la etiqueta
        cv2.rectangle(frame, (int(xmin), int(ymin)), (int(xmax), int(ymax)), (0, 255, 0), 2)
        cv2.putText(frame, f"{label} {confidence:.2f}", (int(xmin), int(ymin) - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)

    # Mostrar el frame con las detecciones
    cv2.imshow("Reconocimiento de Objetos", frame)

    # Esperar a que se presione una tecla
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Liberar recursos
cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

# Rutas a los archivos del modelo
configPath = 'ssd_mobilenet_v3_large_coco_2020_01_14.pbtxt'
weightsPath = 'frozen_inference_graph.pb'

# Cargar el modelo
net = cv2.dnn_DetectionModel(weightsPath, configPath)
net.setInputSize(320, 320)  # Tamaño de entrada del modelo
net.setInputScale(1.0 / 127.5)  # Escala de normalización
net.setInputMean((127.5, 127.5, 127.5))  # Media de normalización
net.setInputSwapRB(True)  # Cambiar de BGR a RGB

# Cargar las etiquetas de las clases
classNames = []
classFile = 'coco.names'  # Archivo con las etiquetas de COCO
with open(classFile, 'rt') as f:
    classNames = f.read().rstrip('\n').split('\n')

# URL del flujo de video RTSP
url_rtsp = "rtsp://admin:Apsa2812@192.168.0.209:554/(h264/MPEG-4)/ch(CANAL)/(main/sub)/av_stream"

# Crear un objeto de captura de video
cap = cv2.VideoCapture(url_rtsp)

# Verificar si la captura de video se ha iniciado correctamente
if not cap.isOpened():
    print("Error al iniciar la captura de video")
    exit()

while True:
    ret, frame = cap.read()
    if not ret:
        print("Error al leer un frame")
        break

    # Obtener dimensiones de la imagen
    height, width, _ = frame.shape

    # Calcular el 60% central de la imagen
    start_x = int(width * 0.2)  # 20% desde el borde izquierdo
    start_y = int(height * 0.2)  # 20% desde el borde superior
    end_x = int(width * 0.8)  # 80% desde el borde izquierdo
    end_y = int(height * 0.8)  # 80% desde el borde superior

    # Recortar el 60% central de la imagen
    cropped_frame = frame[start_y:end_y, start_x:end_x]

    # Detectar objetos en el frame recortado
    classIds, confs, bbox = net.detect(cropped_frame, confThreshold=0.5)

    # Dibujar las detecciones en el frame recortado
    if len(classIds) != 0:
        for classId, confidence, box in zip(classIds.flatten(), confs.flatten(), bbox):
            # Ajustar las coordenadas del cuadro delimitador al frame original
            box[0] += start_x  # Ajustar coordenada x
            box[1] += start_y  # Ajustar coordenada y

            # Dibujar el cuadro delimitador en el frame original
            cv2.rectangle(frame, box, color=(0, 255, 0), thickness=2)
            # Mostrar la etiqueta y la confianza
            label = f"{classNames[classId - 1]}: {confidence:.2f}"
            cv2.putText(frame, label, (box[0] + 10, box[1] + 30),
                        cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2)

    # Mostrar el frame original con las detecciones
    cv2.imshow("Reconocimiento de Objetos", frame)

    # Esperar a que se presione una tecla
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Liberar recursos
cap.release()
cv2.destroyAllWindows()
```
